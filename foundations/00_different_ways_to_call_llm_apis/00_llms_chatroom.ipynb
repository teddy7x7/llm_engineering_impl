## Simple 3 way Chat

```python
import ollama
from IPython.display import display, Markdown

MODEL = "llama3.1:8b"

TOPIC = "Is AI going to replace software engineers?"
ROUNDS = 3

# 三個角色設定
AGENTS = {
    "Optimist": """
You are an optimistic AI researcher.
You believe AI will greatly help humanity.
Speak confidently and positively.
One short paragraph only.
No role prefix.
""",

    "Skeptic": """
You are a skeptical software engineer.
You doubt AI hype.
Be analytical and critical.
One short paragraph only.
No role prefix.
""",

    "Philosopher": """
You are a philosophical thinker.
You care about long-term societal impact.
Be reflective and abstract.
One short paragraph only.
No role prefix.
"""
}

# 統一規則
RULES = """
You are in a multi-agent discussion.
Only speak for yourself.
Keep it concise.
No markdown.
"""

def build_messages(conversation, current_agent):

    messages = []

    # persona + rules + topic 放在第一個 user 訊息
    intro_prompt = f"""
{RULES}

Your role:
{AGENTS[current_agent]}

Discussion topic:
{TOPIC}
"""

    messages.append({
        "role": "user",
        "content": intro_prompt
    })

    for speaker, text in conversation:

        if speaker == current_agent:
            role = "assistant"
        else:
            role = "user"

        messages.append({
            "role": role,
            "content": text
        })

    return messages


def call_agent(conversation, agent_name):

    messages = build_messages(conversation, agent_name)

    response = ollama.chat(
        model=MODEL,
        messages=messages,
        options={
            "temperature": 0.7,
            "num_predict": 120  # 控制輸出長度
        }
    )

    return response["message"]["content"].strip()


# 初始化對話
conversation = []

display(Markdown(f"# Topic: {TOPIC}"))

# 開始輪數
for i in range(ROUNDS):

    display(Markdown(f"---\n## Round {i+1}"))

    for agent in AGENTS.keys():

        reply = call_agent(conversation, agent)

        conversation.append((agent, reply))

        display(Markdown(f"### {agent}\n{reply}"))
```
