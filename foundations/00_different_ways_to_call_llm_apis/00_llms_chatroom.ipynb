````python
import ollama
from IPython.display import display, Markdown

MODEL = "llama3.1:8b"

TOPIC = "Is AI going to replace software engineers?"
ROUNDS = 3

# 三個角色設定
AGENTS = {
    "Optimist": """
You are an optimistic AI researcher.
You believe AI will greatly help humanity.
Speak confidently and positively.
One short paragraph only.
No role prefix.
""",

    "Skeptic": """
You are a skeptical software engineer.
You doubt AI hype.
Be analytical and critical.
One short paragraph only.
No role prefix.
""",

    "Philosopher": """
You are a philosophical thinker.
You care about long-term societal impact.
Be reflective and abstract.
One short paragraph only.
No role prefix.
"""
}

# 統一規則
RULES = """
You are in a multi-agent discussion.
Only speak for yourself.
Keep it concise.
No markdown.
"""

def build_messages(conversation, current_agent):

    messages = []

    # persona + rules + topic 放在第一個 user 訊息
    intro_prompt = f"""
{RULES}

Your role:
{AGENTS[current_agent]}

Discussion topic:
{TOPIC}
"""

    messages.append({
        "role": "user",
        "content": intro_prompt
    })

    for speaker, text in conversation:

        if speaker == current_agent:
            role = "assistant"
        else:
            role = "user"

        messages.append({
            "role": role,
            "content": text
        })

    return messages


def call_agent(conversation, agent_name):

    messages = build_messages(conversation, agent_name)

    response = ollama.chat(
        model=MODEL,
        messages=messages,
        options={
            "temperature": 0.7,
            "num_predict": 120  # 控制輸出長度
        }
    )

    return response["message"]["content"].strip()


# 初始化對話
conversation = []

display(Markdown(f"# Topic: {TOPIC}"))

# 開始輪數
for i in range(ROUNDS):

    display(Markdown(f"---\n## Round {i+1}"))

    for agent in AGENTS.keys():

        reply = call_agent(conversation, agent)

        conversation.append((agent, reply))

        display(Markdown(f"### {agent}\n{reply}"))
````

````python
import ollama
from IPython.display import display, Markdown

MODEL = "llama3.1:8b"

STORY_SETUP = """
情境設定：

某天早上，學校突然在全校廣播中宣布：
「即日起，全面禁止學生訂外送，如違規將記過處分。」

沒有事先說明理由，校園瞬間陷入騷動。
三名學生在午休時於教室展開討論，
並在接下來的事件中，逐漸發現背後真正的原因。

故事必須有明確的起、承、轉、合。
每一輪推進劇情，不可停留在閒聊。
"""
````

````python
AGENTS = {

    "Jay": {
        "identity": "熱血高中生",
        "core_trait": "行動派、情緒外放、極度樂觀",
        "narrative_function": "推動衝突與行動，第一個提出反抗或調查",
        "speaking_style": "語氣活潑、帶驚嘆號、充滿感染力",
    },

    "Klay": {
        "identity": "厭世吐槽型學生",
        "core_trait": "懷疑權威、冷幽默、嘴硬心軟",
        "narrative_function": "質疑事件合理性，揭露矛盾與陰謀感",
        "speaking_style": "語氣冷淡、諷刺、略帶毒舌",
    },

    "Brown": {
        "identity": "理性學霸",
        "core_trait": "冷靜分析、邏輯清晰、觀察細膩",
        "narrative_function": "提供線索與推理，逐步揭露真相",
        "speaking_style": "語氣沉穩、用詞精準、條理清晰",
    }
}
````

````python
def build_persona_prompt(agent_name):

    persona = AGENTS[agent_name]

    return f"""
你正在參與一個三人合作創作的故事。

【你的角色設定】
身份：{persona['identity']}
核心特質：{persona['core_trait']}
劇情功能：{persona['narrative_function']}
說話風格：{persona['speaking_style']}

【故事規則】
- 這是一個連續劇情故事，不是閒聊。
- 每一輪必須推進情節。
- 故事需逐步形成「起承轉合」。
- 僅輸出一個短段落。
- 不要顯示角色名稱。
"""
````

````python
STORY_PHASES = [
    "起：角色對事件做出第一反應，建立衝突。",
    "承：角色開始行動或調查，發現異常線索。",
    "轉：出現重大反轉或隱藏真相。",
    "合：三人理解真相，事件迎來結局。"
]
````

````python
def build_messages(conversation, current_agent, round_index):

    messages = []

    persona_prompt = build_persona_prompt(current_agent)
    phase_instruction = STORY_PHASES[min(round_index, len(STORY_PHASES)-1)]

    intro = f"""
{STORY_SETUP}

【本輪劇情階段】
{phase_instruction}

{persona_prompt}
"""

    messages.append({"role": "user", "content": intro})

    for speaker, text in conversation:

        if speaker == current_agent:
            role = "assistant"
        else:
            role = "user"

        messages.append({
            "role": role,
            "content": text
        })

    return messages
````

````python
def call_agent(
    conversation,
    agent_name,
    round_index,
    model="llama3.1:8b",
    max_history=8,
    temperature=0.7,
    max_tokens=120,
    debug=False
):
    """
    conversation: [(speaker, text), ...]
    agent_name: 當前角色
    round_index: 用於控制劇情階段
    max_history: 限制記憶長度（避免爆 context）
    """

    # 🔹 限制歷史長度（sliding window）
    trimmed_conversation = conversation[-max_history:]

    # 🔹 建構 messages
    messages = build_messages(
        trimmed_conversation,
        agent_name,
        round_index
    )

    try:
        response = ollama.chat(
            model=model,
            messages=messages,
            options={
                "temperature": temperature,
                "num_predict": max_tokens
            }
        )

        content = response.get("message", {}).get("content", "")

        if not content.strip():
            return "[沉默片刻，沒有說話。]"

        # 🔹 防止模型爆寫多段
        content = content.strip()

        # 只保留第一段（保險機制）
        if "\n\n" in content:
            content = content.split("\n\n")[0]

        # 避免模型偷加角色前綴
        for name in AGENTS.keys():
            if content.startswith(name):
                content = content.replace(name, "", 1).strip(":： ")

        if debug:
            print("DEBUG RESPONSE:", response)

        return content

    except Exception as e:
        return f"[發生錯誤: {str(e)}]"
````

````python
# 初始化對話
conversation = []

MAXTOKEN=1000
ROUNDS=20

display(Markdown(f"# Topic: {TOPIC}"))

# 開始輪數
for i in range(ROUNDS):

    display(Markdown(f"---\n## Round {i+1}"))

    for agent in AGENTS.keys():

        reply = call_agent(
            conversation,
            agent,
            round_index=i,
            max_tokens=MAXTOKEN
        )

        conversation.append((agent, reply))

        display(Markdown(f"### {agent}\n{reply}"))
````

````python
import json
import re

def extract_active_characters(text: str) -> list[dict]:
    """
    從混合文字或純 JSON 中擷取「重要角色」的完整資訊。
    
    篩選邏輯：
    - 排除含 exclude_keywords 的角色（旁白、路人）
    - 保留含 include_keywords 的角色（主角、合作夥伴、反派等）
    - 若兩者皆無，預設不納入（可依需求調整）

    Returns:
        重要角色的 dict 列表，每筆包含完整欄位
    """

    INCLUDE_KEYWORDS = {"主角", "合作", "反派", "核心"}
    # EXCLUDE_KEYWORDS = {"旁白", "路人"}
    EXCLUDE_KEYWORDS = {}

    # ── 1. 擷取 JSON 陣列（貪婪模式，避免截斷巢狀結構）──────────────────
    match = re.search(r'\[.+\]', text, re.DOTALL)  # 改為貪婪，確保完整陣列
    if not match:
        print("[警告] 找不到 JSON 陣列，請確認輸入格式")
        return []

    # ── 2. 解析 JSON ──────────────────────────────────────────────────────
    try:
        characters = json.loads(match.group())
    except json.JSONDecodeError as e:
        print(f"[錯誤] JSON 解析失敗：{e}")
        return []

    if not isinstance(characters, list):
        print("[錯誤] JSON 內容不是陣列格式")
        return []

    # ── 3. 篩選重要角色 ───────────────────────────────────────────────────
    active = []

    for char in characters:
        # 跳過非 dict 的異常元素
        if not isinstance(char, dict):
            continue

        function_text = char.get("narrative_function", "")

        # 優先排除
        if any(ex in function_text for ex in EXCLUDE_KEYWORDS):
            continue

        # 符合納入條件才加入
        if any(inc in function_text for inc in INCLUDE_KEYWORDS):
            active.append({
                "name":               char.get("name", "未知"),
                "identity":           char.get("identity", ""),
                "personality":        char.get("personality", ""),
                "narrative_function": function_text,
                "speaking_style":     char.get("speaking_style", ""),
            })

    if not active:
        print("[提示] 未找到符合條件的重要角色，請檢查關鍵字設定")

    return active
````

````python
import json
import re

def extract_all_characters(text: str) -> list[dict]:
    """
    從混合文字或純 JSON 中擷取所有角色的完整資訊。

    Returns:
        所有角色的 dict 列表，每筆包含完整欄位
    """

    # ── 1. 擷取 JSON 陣列 ─────────────────────────────────────────────────
    match = re.search(r'\[.+\]', text, re.DOTALL)
    if not match:
        print("[警告] 找不到 JSON 陣列，請確認輸入格式")
        return []

    # ── 2. 解析 JSON ──────────────────────────────────────────────────────
    try:
        characters = json.loads(match.group())
    except json.JSONDecodeError as e:
        print(f"[錯誤] JSON 解析失敗：{e}")
        return []

    if not isinstance(characters, list):
        print("[錯誤] JSON 內容不是陣列格式")
        return []

    # ── 3. 擷取所有角色 ───────────────────────────────────────────────────
    all_characters = []

    for char in characters:
        if not isinstance(char, dict):
            continue

        all_characters.append({
            "name":               char.get("name", "未知"),
            "identity":           char.get("identity", ""),
            "personality":        char.get("personality", ""),
            "narrative_function": char.get("narrative_function", ""),
            "speaking_style":     char.get("speaking_style", ""),
        })

    if not all_characters:
        print("[提示] 未找到任何角色，請確認 JSON 格式是否正確")

    return all_characters
````

### 🧩 Cell 6️⃣：Step 5 - 修改後的 Multi-Agent 函數

````python
def build_messages(conversation, character_name, story_context):

    messages = []

    intro = f"""
{story_context}

你現在扮演角色：{character_name}

必須依據你的角色設定與目前的劇情進度發言。
一切發言必須以推進劇情發展為原則。

"""
# 如果現在要由旁白先交代整體情況，而非對話內容，則自己的角色不發言。
    
    messages.append({"role": "user", "content": intro})

    for speaker, text in conversation:

        if speaker == character_name:
            role = "assistant"
        else:
            role = "user"

        messages.append({
            "role": role,
            "content": text
        })

    return messages


def call_character(conversation, character_name):

    trimmed = conversation[-MAX_HISTORY:]

    messages = build_messages(trimmed, character_name, STORY_CONTEXT)

    try:
        response = ollama.chat(
            model=MODEL,
            messages=messages,
            options={
                "temperature": TEMPERATURE,
                "num_predict": MAX_TOKENS
            }
        )

        content = response["message"]["content"].strip()

        if "\n\n" in content:
            content = content.split("\n\n")[0]

        return content

    except Exception as e:
        return f"[錯誤: {str(e)}]"
````

````python
def build_messages(conversation, character_name, story_context, can_pass=False):

    pass_instruction = (
        "若你認為此刻不需要發言，請只回覆 PASS，不要加任何其他文字。"
        if can_pass else ""
    )

    intro = f"""
{story_context}

你現在扮演角色：{character_name}

必須依據你的角色設定與目前的劇情進度發言。
一切發言必須以推進劇情往結局發展為原則。
如果現在要由旁白先交代整體情況，而非對話內容，則自己的角色不發言。
{pass_instruction}
"""

    messages = [{"role": "user", "content": intro}]

    for speaker, text in conversation:

        role = "assistant" if speaker == character_name else "user"

        messages.append({
            "role": role,
            "content": f"{speaker}：{text}"  # 加上說話者名稱，讓模型清楚知道是誰說的
        })

    return messages

def call_character(conversation, character_name, can_pass=False):

    trimmed = conversation[-MAX_HISTORY:]

    messages = build_messages(trimmed, character_name, STORY_CONTEXT, can_pass)

    try:
        response = ollama.chat(
            model=MODEL,
            messages=messages,
            options={
                "temperature": TEMPERATURE,
                "num_predict": MAX_TOKENS
            }
        )

        content = response["message"]["content"].strip()

        if "\n\n" in content:
            content = content.split("\n\n")[0]

        # 角色選擇沉默
        if can_pass and content.upper() == "PASS":
            return None

        return content

    except Exception as e:
        return f"[錯誤: {str(e)}]"
````

````python
def call_director(conversation, active_characters, story_context):
    """
    導演模型：根據當前對話，決定下一個應該發言的角色。
    回傳角色名稱字串。
    """
    character_list = "、".join(active_characters)
    
    history = "\n".join([f"{speaker}：{text}" for speaker, text in conversation[-MAX_HISTORY:]])
    
    prompt = f"""
{story_context}

目前對話紀錄：
{history}

現在可以發言的角色有：{character_list}

請根據劇情脈絡，決定下一個最應該發言的角色。
只能回覆角色名稱，不要加任何其他文字。
"""
    
    response = ollama.chat(
        model=MODEL,
        messages=[{"role": "user", "content": prompt}],
        options={"temperature": 0.3, "num_predict": 10}  # 低溫度確保穩定輸出
    )
    
    next_speaker = response["message"]["content"].strip()
    
    # 防止模型回傳不在清單內的角色
    if next_speaker not in active_characters:
        return active_characters[0]  # fallback
    
    return next_speaker


display(Markdown("# 🎬 故事開始"))

conversation.append(('旁白', start_of_the_story))

for i in range(ROUNDS):

    display(Markdown(f"---\n## 第 {i+1} 幕"))

    # 旁白固定開場
    reply = call_character(conversation, '旁白')
    conversation.append(('旁白', reply))
    display(Markdown(f"### 旁白\n{reply}"))

    # 每幕內讓導演決定發言順序，允許多輪對話
    MAX_TURNS_PER_ACT = 5  # 每幕最多幾輪對話
    last_speaker = None

    for _ in range(MAX_TURNS_PER_ACT):

        next_speaker = call_director(conversation, ACTIVE_CHARACTERS, STORY_CONTEXT)

        # 避免同一角色連續發言
        if next_speaker == last_speaker:
            continue

        reply = call_character(conversation, next_speaker)
        conversation.append((next_speaker, reply))
        display(Markdown(f"### {next_speaker}\n{reply}"))

        last_speaker = next_speaker
````
